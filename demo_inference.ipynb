## 3. Pre-Cache Models & Weights via Hugging Face Hub
Downloads base Stable Diffusion 1.5, AnimateDiff Motion Module v2, and IP-Adapter into the local Hugging Face cache for rapid startup.

```python
from huggingface_hub import snapshot_download, hf_hub_download

print("Pre-caching base models...")
# 1. AnimateDiff Motion Adapter v2
snapshot_download(repo_id="guoyww/animatediff-motion-adapter-v1-5-2")

# 2. IP-Adapter SD 1.5 weights
hf_hub_download(repo_id="h94/IP-Adapter", subfolder="models", filename="ip-adapter_sd15.bin")

# 3. CLIP ViT-H Image Encoder
snapshot_download(repo_id="laion/CLIP-ViT-H-14-laion2B-s32B-b79K")

print("All foundational weights cached successfully!")
```
